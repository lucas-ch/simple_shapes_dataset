```python
import numpy as np
from pathlib import Path

from simple_shapes_dataset.cli.utils import generate_dataset, generate_dataset_biased, generate_image,generate_color
from simple_shapes_dataset.cli import create_dataset_biased
from tqdm import tqdm
from simple_shapes_dataset.version import __version__

```

```python
class_configs = {
    0: {'fixed_color_hls': np.array([60, 128, 255])},   # vert  = cercle
    1: {'fixed_color_hls': np.array([0, 128, 255])},    # rouge = triangle
    2: {'fixed_color_hls': np.array([120, 128, 255])},  # bleu  = carré
}

color_given_class = {
    #  vert   rouge  bleu
    0: [0.3,  0.3,  0.4],  # diamant
    1: [0.8,  0.1,  0.1],  # oeuf
    2: [0.2,  0.7,  0.1],  # triangle
}
```

```python
color_given_class_0 = {
    #  vert   rouge  bleu
    0: [0,  0,  1],  # diamant
    1: [1,  0, 0],  # oeuf
    2: [0,  1,  0],  # triangle
}


dataset = generate_dataset_biased(
    n_samples= 30,
    class_configs = class_configs,
    min_scale=7,
    max_scale=14,
    min_lightness=46,
    max_lightness=256,
    fixed_color_rate=0.1,
    color_given_class=  color_given_class_0,
    imsize=32)
```

```python
%matplotlib inline

from typing import cast
from matplotlib import pyplot as plt
from matplotlib.axes import Axes

dpi = 1
imsize = 32
enumerator = tqdm(
        enumerate(
            zip(
                dataset.classes,
                dataset.locations,
                dataset.sizes,
                dataset.rotations,
                dataset.colors,
                strict=True,
            )
        ),
        total=len(dataset.classes),
    )

for k, (cls, location, size, rotation, color) in enumerator:
    # Création de la figure
    fig, ax = plt.subplots(figsize=(imsize / dpi, imsize / dpi), dpi=dpi)
    ax = cast(plt.Axes, ax)
    generate_image(ax, cls, location, size, rotation, color, imsize)
    ax.set_facecolor("black")    
    plt.show()
    plt.close(fig)
```
